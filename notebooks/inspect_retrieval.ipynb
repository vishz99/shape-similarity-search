# Inspect retrieval

Query an existing part by stem and view the top 3 matches with thumbnails.

Run this notebook with the `brepnet` conda environment as its kernel.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
from PIL import Image

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from query_similar_designs import query_by_stem

STEP_DIR = PROJECT_ROOT / "data/processed/fusion360_subset"
```

```python
query_stem = "100243_9fb796fe_0006"
top_k = 3

results = query_by_stem(query_stem, top_k=top_k)
results
```

```python
fig, axes = plt.subplots(1, top_k + 1, figsize=(4 * (top_k + 1), 4))

query_img = Image.open(STEP_DIR / f"{query_stem}.png")
axes[0].imshow(query_img)
axes[0].set_title(f"Query\n{query_stem}")
axes[0].axis("off")

for ax, (stem, dist) in zip(axes[1:], results):
    img = Image.open(STEP_DIR / f"{stem}.png")
    ax.imshow(img)
    ax.set_title(f"{stem}\ndistance={dist:.4f}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
